```python
from hcc_release_paths import project_path as _hcc_path
# ============================================================
# Step7 GSEA
# 初始化
# ============================================================

import os
import pandas as pd
import numpy as np


project_dir = _hcc_path('')


step5_dir = os.path.join(
    project_dir,
    "LIHC_RNA_Step5_DEG"
)


step7_dir = os.path.join(
    project_dir,
    "LIHC_RNA_Step7_GSEA"
)


result_dir = os.path.join(
    step7_dir,
    "Results"
)


fig_dir = os.path.join(
    step7_dir,
    "Figures"
)


for d in [
    step7_dir,
    result_dir,
    fig_dir
]:
    os.makedirs(
        d,
        exist_ok=True
    )


print(step7_dir)
```

```python
deg_file = os.path.join(
    step5_dir,
    "DEG_Result",
    "LIHC_Tumor_vs_Normal_DEG_All.csv"
)


deg = pd.read_csv(
    deg_file
)


print(
    deg.shape
)


deg.head()
```

```python
# ============================================================
# 构建rank文件
# ============================================================


gsea_rank = deg[
    [
        "Gene",
        "log2FoldChange"
    ]
].copy()


gsea_rank = (
    gsea_rank
    .dropna()
    .sort_values(
        by="log2FoldChange",
        ascending=False
    )
)


print(
    gsea_rank.head()
)


print(
    gsea_rank.shape
)
```

```python
gsea_rank.to_csv(
    os.path.join(
        result_dir,
        "LIHC_GSEA_rank.rnk"
    ),
    sep="\t",
    index=False,
    header=False
)


print("rank file saved")
```

```python
hallmark = gp.get_library(
    name="MSigDB_Hallmark_2020",
    organism="Human"
)


print(
    "Hallmark pathways:",
    len(hallmark)
)


list(hallmark.keys())[:10]
```

```python
import os


gsea_results = gp.prerank(

    rnk=os.path.join(
        result_dir,
        "LIHC_GSEA_rank.rnk"
    ),

    gene_sets=hallmark,

    permutation_num=1000,

    seed=123,

    outdir=None,

    verbose=True
)


print("GSEA finished")
```

```python
gsea_table = gsea_results.res2d


print(
    gsea_table.shape
)


gsea_table.head(10)
```

```python
# ============================================================
# Save GSEA results
# ============================================================

gsea_table.to_csv(
    os.path.join(
        result_dir,
        "LIHC_Hallmark_GSEA_results.csv"
    ),
    index=False
)


print("GSEA results saved")
```

```python
# ============================================================
# GSEA NES dotplot save version
# ============================================================

import matplotlib.pyplot as plt
import numpy as np
import os


gsea_plot = (
    gsea_table
    .sort_values(
        by="NES",
        key=lambda x: abs(x),
        ascending=False
    )
    .head(20)
)


fig, ax = plt.subplots(
    figsize=(9,8)
)


ax.scatter(
    gsea_plot["NES"],
    range(len(gsea_plot)),
    s=(
        -np.log10(
            gsea_plot["FDR q-val"] + 1e-300
        )
        *50
    )
)


ax.set_yticks(
    range(len(gsea_plot))
)


ax.set_yticklabels(
    gsea_plot["Term"]
)


ax.axvline(
    0,
    linestyle="--"
)


ax.set_xlabel(
    "Normalized Enrichment Score (NES)"
)


ax.set_ylabel(
    "Hallmark Pathway"
)


ax.set_title(
    "Hallmark GSEA in LIHC"
)


ax.invert_yaxis()


plt.tight_layout()


# 先保存
fig.savefig(
    os.path.join(
        fig_dir,
        "LIHC_Hallmark_GSEA_NES_dotplot.png"
    ),
    dpi=600,
    bbox_inches="tight"
)


plt.show()
```
