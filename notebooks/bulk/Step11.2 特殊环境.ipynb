```python
from hcc_release_paths import project_path as _hcc_path
# ============================================================
# Step11.9 ESTIMATE
# Run in Python (ESTIMATE) kernel
# ============================================================

import os
import pandas as pd

from TMEImmune import TME_score


project_dir = _hcc_path('')

step11_dir = os.path.join(
    project_dir,
    "LIHC_RNA_Step11_Immune"
)

input_dir = os.path.join(
    step11_dir,
    "Input"
)

result_dir = os.path.join(
    step11_dir,
    "Results"
)


estimate_input = os.path.join(
    input_dir,
    "LIHC_ESTIMATE_Input_365.tsv"
)


rna_estimate = pd.read_csv(
    estimate_input,
    sep="\t",
    index_col=0
)


print("ESTIMATE input shape:")
print(rna_estimate.shape)

display(
    rna_estimate.iloc[:5, :5]
)
```

```python
# ============================================================
# Step11.10 Run ESTIMATE
# ============================================================

estimate_score = TME_score.get_score(
    rna_estimate,
    method=["ESTIMATE"]
)


print("ESTIMATE shape:")
print(estimate_score.shape)

print("\nColumns:")
print(estimate_score.columns.tolist())

display(
    estimate_score.head()
)
```

```python
# ============================================================
# Step11.11 Save ESTIMATE scores
# ============================================================

estimate_output = os.path.join(
    result_dir,
    "LIHC_ESTIMATE_Scores_365.csv"
)


estimate_score.to_csv(
    estimate_output
)


print("Saved:")
print(estimate_output)
```
