```python
from hcc_release_paths import project_path as _hcc_path
import pandas as pd
import numpy as np
import os


project_dir = _hcc_path('')


step5_dir = os.path.join(
    project_dir,
    "LIHC_RNA_Step5_DEG"
)


os.makedirs(
    step5_dir,
    exist_ok=True
)


print(step5_dir)
```

```python
from hcc_release_paths import project_path as _hcc_path
normal_file = _hcc_path('TCGA-LIHC/processed/Normal_RawCounts.tsv')


normal = pd.read_csv(
    normal_file,
    sep="\t"
)


print(normal.shape)

normal.head()
```

```python
normal = normal.set_index(
    "Gene_Symbol"
)


print(normal.shape)
```

```python
# 保留至少10个normal样本表达>0


normal_filtered = normal[
    (normal>0).sum(axis=1)>=10
]


print(normal_filtered.shape)
```

```python
tumor = pd.read_csv(
    os.path.join(
        project_dir,
        "LIHC_RNA_Step1",
        "LIHC_RNA_FilteredCounts_371.tsv"
    ),
    sep="\t",
    index_col=0
)


print(tumor.shape)
```

```python
common_genes = (
    tumor.index
    .intersection(
        normal_filtered.index
    )
)


print(
    len(common_genes)
)
```

```python
tumor_common = tumor.loc[
    common_genes
]


normal_common = normal_filtered.loc[
    common_genes
]


deg_matrix = pd.concat(
    [
        tumor_common,
        normal_common
    ],
    axis=1
)


print(
    deg_matrix.shape
)
```

```python
deg_matrix.to_csv(
    os.path.join(
        step5_dir,
        "LIHC_Tumor_Normal_Counts.csv"
    )
)


print("Saved")
```

```python
from hcc_release_paths import project_path as _hcc_path
import os


project_dir = _hcc_path('')


step5_dir = os.path.join(
    project_dir,
    "LIHC_RNA_Step5_DEG"
)


normal_dir = os.path.join(
    step5_dir,
    "Normal_RNA"
)


result_dir = os.path.join(
    step5_dir,
    "DEG_Result"
)


fig_dir = os.path.join(
    step5_dir,
    "Figures"
)


for d in [
    step5_dir,
    normal_dir,
    result_dir,
    fig_dir
]:
    os.makedirs(
        d,
        exist_ok=True
    )


print(step5_dir)
```

```python
import pandas as pd
import numpy as np


normal_file = os.path.join(
    normal_dir,
    "LIHC_Normal_Counts.tsv"
)


normal = pd.read_csv(
    normal_file,
    sep="\t"
)


print(normal.shape)

normal.head()
```

```python
normal = normal.set_index(
    "Gene_Symbol"
)


print(normal.shape)
```

```python
# ============================================================
# Step5.6 读取Tumor counts
# ============================================================


tumor_file = os.path.join(
    project_dir,
    "LIHC_RNA_Step1",
    "LIHC_RNA_FilteredCounts_371.tsv"
)


tumor = pd.read_csv(
    tumor_file,
    sep="\t",
    index_col=0
)


print(tumor.shape)

tumor.head()
```

```python
# ============================================================
# Step5.8 common genes
# ============================================================


common_genes = (
    tumor.index
    .intersection(
        normal.index
    )
)


print(
    "Common genes:",
    len(common_genes)
)
```

```python
tumor_common = tumor.loc[
    common_genes
]


normal_common = normal.loc[
    common_genes
]


print(
    tumor_common.shape
)


print(
    normal_common.shape
)
```

```python
# ============================================================
# Step5.10 合并
# ============================================================


deg_counts = pd.concat(
    [
        tumor_common,
        normal_common
    ],
    axis=1
)


print(
    deg_counts.shape
)
```

```python
deg_counts.to_csv(
    os.path.join(
        step5_dir,
        "LIHC_Tumor_Normal_Counts.csv"
    )
)


print("Merged counts saved")
```
